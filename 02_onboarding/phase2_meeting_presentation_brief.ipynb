# Phase 2 Meeting Brief

## 1. Meeting Goal

- Align the selected starting members on Phase 2 expectations.
- Explain what a Phase 2-aligned task should look like.
- Explain how trainers should think before creating a task.
- Explain what QA must validate before approval.
- Make sure every task shows a real, defensible single-agent vs multi-agent gap.
- Make sure the team avoids the rejection patterns seen in Phase 1.

---

## 2. Main Phase 2 Message

- Phase 2 is not just more Phase 1 task creation.
- The client expects tasks that are:
  - more realistic
  - more agentic
  - more multi-stage
  - more externally grounded or tool-diverse
  - more defensible during client QA
- The gap must be real.
- The gap should not come from:
  - artificial traps
  - hidden restrictions
  - strict-match issues
  - Judge LLM hallucination
  - verifier weakness
  - reward hacking
  - packaging or execution mistakes

---

## 3. Mandatory Baseline for Every Phase 2 Task

Every Phase 2 task must satisfy these baseline expectations.

### 3.1 Long-Horizon Multi-Stage Workflow

- This is required for all Phase 2 tasks.
- The task should naturally require multiple stages, such as:
  - collect evidence
  - verify evidence
  - analyze or classify information
  - reconcile conflicts
  - produce a final artifact
  - summarize risks or findings
- These stages do not need to be explicitly listed in `instruction.md`.
- Stages should be included in the `decomposition.yaml` as sub agents.
- `instruction.md` should describe what the task needs to solve and the goal, it should not be looks like AI generated or very long prompt.
- We should not write the full solving process inside `instruction.md`. instead we need to make `instruction.md` more human like prompt.

Example of good framing:

- Ask the agent to create a final evidence-backed report, spreadsheet, deck, or analysis.
- Include clear output requirements.
- Include real constraints and expected evidence.

Avoid:

- Writing the internal stages as step-by-step instructions.
- Over-explaining the intended solving strategy.

---

### 3.2 Specialist or Hierarchical Coordination

- Phase 2 tasks should benefit from different roles or layers of agents.
- Avoid simple flat fan-out patterns only.
- Weak pattern:
  - “Split 50 files across 50 workers.”
- Stronger patterns:
  - researcher wave → verifier wave → synthesis wave
  - legal specialist + data specialist + citation specialist + reducer
  - shared evidence collection → parallel analysis branches → final report
  - manager → sub-manager → worker → verifier → reducer
- A manager-spawns-subagents pattern is preferred where it helps structure the task.

---

### 3.3 Real-World or Realistic Task Design

- The task should feel like a real-world work request.
- The dataset or sources should be real and externally verifiable.
- Avoid synthetic tasks without clear provenance.
- Avoid AI-generated-looking prompts.
- Avoid artificially complex test cases that do not represent realistic work.
- If live fetching is unreliable, hosted/pinned sources can be used:
  - Drive
  - S3
  - GitHub release
  - public docs
  - public PDFs

---

### 3.4 Rubric-Based Evaluation

- Evaluation should be based on visible requirements.
- Do not depend mainly on a heavy hand-built Oracle.
- Oracle creation increases AHT, so Phase 2 should avoid unnecessary Oracle-heavy tasks.
- Good rubric checks include:
  - Are citations real?
  - Are claims supported by the cited source?
  - Are required sections present?
  - Are calculations internally consistent?
  - Does each row have evidence?
  - Does the evidence link prove the row?
  - Did the agent clearly mark missing information?
  - Does the output follow the required rules?
- LLM judge can be used, but preferably after structural checks.
- Make sure the LLM is not hallucinating the judge result.

---

### 3.5 Clear Single-Agent Failure Mode

- The single-agent failure mode must be natural and defensible.
- Single agent should struggle because the task is genuinely:
  - broad
  - multi-source
  - multi-stage
  - verification-heavy
  - artifact-heavy
  - coordination-heavy
- The gap should not be created by:
  - artificial strict-match traps
  - hidden information
  - mode-specific instructions
  - single-agent-only restrictions
  - formatting-only failures
  - parser failures
  - timeout/latency artifacts

---

### 3.6 Trainer and QA Must Validate the True Gap

- Trainers must debug their own task.
- QA must independently debug the same task.
- The goal is to confirm whether the task achieved a true gap.
- Do not blindly trust Judge LLM.
- Judge LLM can hallucinate or misjudge.
- Client saw cases where Judge LLM gave a lower score, but the client found the single-agent output and Oracle score were actually much higher.
- This means trainer and QA must catch Judge LLM hallucination before client submission.

Trainer and QA should check:

- single-agent output
- multi-agent output
- rubric
- verifier behavior
- client expectation
- whether the gap is real
- whether the gap is caused by a valid task difficulty

---

## 4. Select at least 1 option of the following Phase 2 Classes

Each task should pick at least one client-aligned optional class.

### 4.1 Browsing / External Retrieval

Use this class when the task requires the agent to fetch or verify information externally.

Examples:

- web sources
- public portals
- APIs
- hosted files
- public docs
- public PDFs
- regulatory filings
- procurement portals
- source verification through links

Important notes:

- Browsing is a strong Phase 2 option.
- A task that only reads local files may look too similar to Phase 1.
- If using specific URLs, decomposition/instructions should constrain the agents to the intended URLs.
- You can ask the model to search for specific information on the internet and provide evidence.
---

### 4.2 R-4 Long Writing

Use this class when the final output requires substantial written analysis.

Examples:

- detailed report
- legal briefing
- risk memo
- compliance analysis
- policy summary
- research findings
- evidence-backed recommendations
- long written summaries based on links or files

Important notes:

- Long writing should depend on evidence, not generic prose.
- The writing should show analysis, synthesis, and reasoning.
- A spreadsheet alone should not automatically count as long writing.
- A spreadsheet plus a substantial written report can count as long writing.

---

### 4.3 R-7 Creative Viz

Use this class when the task requires visual communication of findings.

Examples:

- slide deck
- presentation
- charts
- dashboard-style output
- timeline
- risk matrix
- visual evidence map
- comparison visuals
- creative visualization based on research findings

Important notes:

- Presentation / slide generation should be counted under Creative Viz.
- Creative Viz should not be decoration only.
- The visualization should help communicate the task findings.

---

### 4.4 Spreadsheet / Structured Output

Use this class when the task requires structured row/column outputs.

Examples:

- `output.xlsx`
- evidence matrix
- classification table
- audit sheet
- scoring sheet
- risk register
- row-level source verification
- structured flags and notes

Important notes:

- Spreadsheet / structured output is separate from Long Writing.
- It can support Long Writing, but it does not automatically fall under Long Writing.
- It is useful when combined with evidence checks and rubric scoring.

---

## 5. What Strong Phase 2 Tasks Should Look Like

A strong Phase 2 task should usually include:

- a realistic user-facing objective
- real-world or realistic data
- one or more external or tool-diverse components
- specialist or hierarchical agent coordination
- multiple stages
- evidence collection and verification
- analysis or classification
- conflict reconciliation where relevant
- final artifact generation
- rubric-based scoring
- a clear reason why single-agent should struggle
- a clear reason why multi-agent should do better

---

## 6. What to Avoid in Phase 2

Avoid tasks that are mainly:

- local file reading only
- simple fan-out only
- Phase 1-style extraction with no new pattern
- synthetic-looking without provenance
- strict exact-match traps
- benchmark-contaminated wording
- hidden implementation hints
- mode-specific instructions
- single-agent-only restrictions
- reward hacking
- format-only scoring
- parser-failure scoring
- all-or-nothing scoring without meaningful partial credit
- weak verifier checks that do not test the intended capability
- tasks where single-agent already scores too high because the task is too easy

---

## 7. How to Think About `instruction.md`

`instruction.md` should include:

- the task objective
- the expected final deliverable
- source or artifact requirements
- output format requirements
- evidence requirements
- constraints the user needs followed
- quality expectations

`instruction.md` should not include:

- the full internal solving stages
- the intended subagent decomposition
- hidden verifier logic
- gold answers
- formulas that leak the answer
- multi-agent-only guidance
- benchmark or gap-widening language

Key point:

- The task should require a multi-stage workflow naturally.
- We do not need to explicitly tell the single agent or multi-agent how to solve it stage by stage.

---

## 8. Decomposition Guidance

Good decomposition should:

- assign clear roles
- avoid unnecessary flat fan-out
- use hierarchy where useful
- include verifier/corrective agents when needed
- avoid leaking answers or verifier hints
- avoid MA-only implementation guidance that explains the gap
- constrain source usage when specific sources are required
- include fallback behavior where possible
- avoid waiting for every subagent if partial output is acceptable
- avoid making the reducer run only after a long sequential bottleneck

Lessons from the shared task discussion:

- If the orchestrator spawns 50 subagents sequentially, timeout risk increases.
- If subagents browse freely, they may follow wrong or invalid links.
- If subagents return empty results silently, the orchestrator may not adapt.
- If the reducer never runs, the final artifact may not be written and the score can become zero.
- First validate the task design before blaming OpenCode or tool issues.

---

## 9. Verifier and Rubric Guidance

Verifier should test the actual intended capability.

Good verifier/rubric checks:

- required files exist
- required sheets or sections exist
- required rows or items are present
- claims have evidence
- URLs are real
- URLs support the claim
- calculations are internally consistent
- missing information is clearly marked
- output follows rules
- final artifact is complete

Avoid verifier issues such as:

- rewarding only structure over accuracy
- exact phrase dependency
- exact oracle wording dependency
- strict parser fail-closed before semantic judging
- binary all-or-nothing gates without partial credit
- score caps or multipliers that create the gap unfairly
- double penalty for the same error
- verifier denominator drift between runs
- reward recording corruption

Important:

- If single-agent gets a high score, check whether the task is actually testing the right capability.
- If the verifier gives a surprising result, debug the verifier.
- If Judge LLM gives a surprising result, manually inspect the outputs.

---

## 10. Phase 1 Client-Rejection Risks to Avoid

These are known client-risk patterns that can lead to rejection when they affect the measured SA-MA gap or would likely trigger client rejection.

Known risk patterns:

- explicit gap wideners or single-agent-only restrictions
- prompt/information asymmetry between single-agent and multi-agent runs
- benchmark or multi-agent wording in the instruction
- decomposition leakage
- hidden answers
- hidden formulas
- verifier hints
- MA-only guidance that explains the gap
- score caps, ceilings, floors, multipliers, or mode-specific reward branches
- binary gates or all-or-nothing scoring without meaningful partial credit
- verifier rewarding structure, schema, word count, or format over accuracy
- peripheral ask gap where a secondary requirement creates the measured gap
- stale logs or task-state mismatch
- verifier denominator or test-count drift across runs
- Oracle/gold output invalid against `instruction.md`
- Oracle-copy or gold-artifact contamination in real runs
- timeout/latency false gap
- reward recording or result-writing corruption
- fail-closed infrastructure, parser, or LLM judge failure
- strict parser fail-closed before semantic judging
- exact phrase, exact substring, or oracle wording dependency
- double penalty for the same error
- duplicate or near-variant task
- mislabeled packaging or wrong-task artifacts
- Throttling output tokens/second low for single agent cause timeout.

---

## 11. How Trainers Should Debug Their Own Task

Before submission, each trainer should ask:

- Did I create a realistic Phase 2 task?
- Does it have a natural long-horizon multi-stage workflow?
- Does it use at least one optional Phase 2 class?
- Is the dataset/source real or realistic?
- Is the source provenance clear?
- Does the task require specialist or hierarchical coordination?
- Is the single-agent failure mode natural?
- Is the multi-agent improvement explainable?
- Is the verifier testing the actual capability?
- Is the scoring fair?
- Is there meaningful partial credit?
- Did Judge LLM hallucinate or misjudge anything?
- Did I manually inspect the single-agent and multi-agent outputs?
- Would the client see the gap as valid?

---

## 12. Pilot Execution Expectations

For the pilot group:

- Start gradually with selected high-quality contributors.
- Each contributor should document the creative idea before full execution.
- The idea should be reviewed before the trainer spends time building the full task.
- Early review should confirm:
  - Phase 2 alignment
  - optional class selected
  - realistic task design
  - source/tool plan
  - expected single-agent failure mode
  - expected multi-agent advantage
  - evaluation/rubric plan
  - risks or blockers
